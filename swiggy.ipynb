## Data Cleaning

```python
for i in df.index:
    if ('mins' in df.loc[i , 'rating'] ):
        df.loc[i, 'time_minutes']=  df.loc[i , 'rating']

df.loc[::, 'time_minutes']
```

```python
df['rating'] = df['rating'].str.replace('mins','')
```

```python
df['time_minutes'] = df['time_minutes'].str.replace('mins','')
```

```python
df['time_minutes'].sample(10)
```

```python
s = df['time_minutes'].astype(str).str.strip().str.split('-', n=1, expand=True)

low = pd.to_numeric(s[0], errors='coerce')
high = pd.to_numeric(s[1], errors='coerce') if 1 in s.columns else low

df['time_minutes'] = (low + high.fillna(low)) / 2
```

```python

df['rating'].isnull().sum()
df['rating'].unique()
```

```python
df['rating'] = df['rating'].astype('f')
df['rating'].unique()
```

```python
for i in df['rating']:
    if (i > 5):
        df.loc[df['rating'] == i, 'rating'] = np.nan
```

```python
df['rating'] = df['rating'].fillna(df['rating'].mean()) 
```
